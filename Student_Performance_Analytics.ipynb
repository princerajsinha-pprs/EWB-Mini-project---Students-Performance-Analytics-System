***Subjects define***

```python
subjects = [
    "Python",
    "NumPy",
    "Pandas",
    "Statistics",
    "Data_Analysis"
]

print("Subjects:", subjects)
```

***Total Marks Function***

```python
def calculate_total(marks):
    return int(np.sum(marks))
```

```python
marks = [78, 85, 74, 88, 82]

print("Total Marks:", calculate_total(marks))
```

***Average Function***

```python
def calculate_average(marks):
    return round(float(np.mean(marks)), 2)
```

```python
print("Average Marks:", calculate_average(marks))
```

***Grade Function***

```python
def assign_grade(average):

    if average >= 90:
        return "A+"

    elif average >= 80:
        return "A"

    elif average >= 70:
        return "B"

    elif average >= 60:
        return "C"

    elif average >= 50:
        return "D"

    else:
        return "F"
```

```python
print(assign_grade(93))
print(assign_grade(81))
print(assign_grade(65))
```

***Pass/Fail Function***

```python
def check_pass(marks, attendance):

    for mark in marks:

        if mark < 40:
            return "Fail"

    if attendance < 75:
        return "Fail"

    return "Pass"
```

```python
print(check_pass([78,85,74,88,82], 91))
print(check_pass([55,61,58,64,60], 72))
```

***Student Processing***

```python
totals = []
averages = []
grades = []
results = []

for index, row in df.iterrows():

    marks = [
        row["Python"],
        row["NumPy"],
        row["Pandas"],
        row["Statistics"],
        row["Data_Analysis"]
    ]

    total = calculate_total(marks)
    average = calculate_average(marks)
    grade = assign_grade(average)
    result = check_pass(marks, row["Attendance"])

    totals.append(total)
    averages.append(average)
    grades.append(grade)
    results.append(result)
```

***Result added in data frame***

```python
df["Total"] = totals
df["Average"] = averages
df["Grade"] = grades
df["Result"] = results

df.head()
```

***Subject-wise Analysis***

```python
subject_analysis = []

for subject in subjects:

    subject_analysis.append({
        "Subject": subject,
        "Average": round(df[subject].mean(), 2),
        "Highest": df[subject].max(),
        "Lowest": df[subject].min()
    })

subject_df = pd.DataFrame(subject_analysis)

subject_df
```

***Best Subject***

```python
best_subject = subject_df.loc[
    subject_df["Average"].idxmax()
]

print("Best Subject:", best_subject["Subject"])
print("Average Marks:", best_subject["Average"])
```
